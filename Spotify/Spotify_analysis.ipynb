```python
def print_section_header(title, width=70):
    """Print a formatted section header."""
    print("=" * width)
    print(title)
    print("=" * width)
```

```python
def summarize_dataframe(df):
    """Print summary statistics for the dataframe."""
    print_section_header("SUMMARY STATISTICS")
    print(df.describe())
```

```python
def check_duplicates(df, column="Artist Name"):
    """Report duplicate rows and duplicate values in a given column."""
    print(f"\nDuplicate rows: {df.duplicated().sum()}")
    print(f"Duplicate {column.lower()}: {df[column].duplicated().sum()}")
```

```python
def find_outliers_iqr(df, column, multiplier=1.5):
    """Identify outliers in a column using the IQR method."""
    q1, q3 = df[column].quantile([0.25, 0.75])
    iqr = q3 - q1
    upper_bound = q3 + multiplier * iqr
    return df[df[column] > upper_bound]
```

```python
def report_outliers(df, column="Total Streams (in millions)", top_n=10):
    """Print outlier summary and top offenders for a given column."""
    outliers = find_outliers_iqr(df, column)
    print(f"\nOutliers ({column}, IQR method): {len(outliers)} artists")
    print(
        outliers[["Artist Name", column]]
        .head(top_n)
        .to_string(index=False)
    )
```
