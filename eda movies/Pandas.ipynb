```python
import pandas as pd
```

```python
df = pd.read_csv('Movie_Data (1).csv')
df
```

Task 1

```python
highest_box_office = df['box_office'].max()
top_movie = df[df['box_office'] == highest_box_office]
print(top_movie[['title', 'genre','rating','release_year']])

Average_rating = df['rating'].mean()
print('\nAverage_rating:',Average_rating)

Median = df['release_year'].max(),  df['genre'].mode()
print('\nMedian and mode:',Median)
```

Task 2

```python
Rating = df.loc[df['rating']>7]
print("Rating")
print(Rating)

top_10_box_office = df.sort_values(by='box_office', ascending=False).head(10)
print("\nTop 10 box office")
print(top_10_box_office)

df['Profitability'] = df['box_office']/100_000_000
print('\nProfitability')
print(df[['title', 'box_office', 'Profitability']])
```
